```python
import torch
from takumi import masking
```

```python
masker = masking.Corruptor(
    rate=masking.rates.BetaRate(mean=0.20, concentration=1),
    select=masking.selectors.Bernoulli(),
    corrupt=masking.schemes.MaskOnly(mask_token_id=3),
    special_token_ids=[0, 1, 2, 3]
)
```

```python
x = torch.randint(4, 100, (1, 100))
```

```python
counter = 0
n = 10000
for _ in range(n):
    output = masker(x)
    counter += output.masked.sum() / output.masked.shape[-1]


print(counter / n)
```

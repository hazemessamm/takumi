```python
import torch
from takumi import masking
```

```python
masker = masking.Corruptor(
    rate=masking.rates.ScheduleRate(total_timesteps=1, schedule="cosine", time="continuous"),
    select=masking.selectors.Bernoulli(),
    corrupt=masking.schemes.MaskOnly(mask_token_id=3),
    special_token_ids=[0, 1, 2, 3]
)
```

```python
x = torch.randint(4, 100, (2, 100))
```

```python
counter = 0
n = 10000
for _ in range(n):
    output = masker(x, seed=4, timestep=[10, 20])
    counter += output.masked.sum() / output.masked.shape[-1]


print(counter / n)
```

```python
output = masker(x, seed=0, timestep=[0.1, 0.2])
```
